```python
dbutils.widgets.text("catalog", "dbr_dev_ua5816bd")
dbutils.widgets.text("schema_prefix", "elina_sharabura")

CATALOG = dbutils.widgets.get("catalog")
SCHEMA_PREFIX = dbutils.widgets.get("schema_prefix")

BRONZE_SCHEMA = f"{CATALOG}.{SCHEMA_PREFIX}_bronze"
SILVER_SCHEMA = f"{CATALOG}.{SCHEMA_PREFIX}_silver"
```

```python
spark.sql(f"CREATE TABLE IF NOT EXISTS {BRONZE_SCHEMA}.customers")
spark.sql(f"CREATE TABLE IF NOT EXISTS {BRONZE_SCHEMA}.products")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {SILVER_SCHEMA}.customers (
    customer_sk         BIGINT GENERATED ALWAYS AS IDENTITY,
    customer_id         INT       NOT NULL,
    name                STRING,
    email               STRING,
    country             STRING,
    age                 INT,
    signup_date         DATE,
    marketing_opt_in    BOOLEAN,
    record_hash         STRING    NOT NULL,
    effective_from      TIMESTAMP NOT NULL,
    effective_to        TIMESTAMP,
    is_current          BOOLEAN   NOT NULL,
    source_filename     STRING,
    ingestion_timestamp TIMESTAMP,
    processed_at        TIMESTAMP NOT NULL
)
CLUSTER BY (customer_id)
""")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {SILVER_SCHEMA}.products (
    product_id          INT       NOT NULL,
    category            STRING,
    name                STRING,
    price_usd           DOUBLE,
    cost_usd            DOUBLE,
    margin_usd          DOUBLE,
    record_hash         STRING    NOT NULL,
    source_filename     STRING,
    ingestion_timestamp TIMESTAMP,
    processed_at        TIMESTAMP NOT NULL
)
CLUSTER BY (product_id)
""")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {SILVER_SCHEMA}.dq_quarantine (
    source_table   STRING        NOT NULL,
    record         STRING        NOT NULL,
    failed_rules   ARRAY<STRING> NOT NULL,
    quarantined_at TIMESTAMP     NOT NULL
)
""")
```
