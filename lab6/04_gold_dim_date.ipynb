```python
dbutils.widgets.text("catalog", "dbr_dev_ua5816bd")
dbutils.widgets.text("schema_prefix", "elina_sharabura")
dbutils.widgets.text("start_date", "2020-01-01")
dbutils.widgets.text("end_date", "2030-12-31")

CATALOG = dbutils.widgets.get("catalog")
SCHEMA_PREFIX = dbutils.widgets.get("schema_prefix")
START_DATE = dbutils.widgets.get("start_date")
END_DATE = dbutils.widgets.get("end_date")

DIM_DATE_TABLE = f"{CATALOG}.{SCHEMA_PREFIX}_gold.dim_date"
```

```python
spark.sql(f"""
INSERT OVERWRITE {DIM_DATE_TABLE}
    (date_key, full_date, year, quarter, month, month_name, day_of_month, day_name, is_weekend)
SELECT
    CAST(date_format(d, 'yyyyMMdd') AS INT),
    d,
    year(d),
    quarter(d),
    month(d),
    date_format(d, 'MMMM'),
    dayofmonth(d),
    date_format(d, 'EEEE'),
    dayofweek(d) IN (1, 7)
FROM (
    SELECT explode(sequence(DATE '{START_DATE}', DATE '{END_DATE}', INTERVAL 1 DAY)) AS d
)
""")
```
